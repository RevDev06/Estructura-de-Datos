```python
# capturar el tamaño de la matriz
i = int(input("Ingresa el numero de filas: "))
j = int(input("Ingresa el numero de columnas: "))

mat = []
for fila in range(i):
    fila_mat = []
    for columna in range(j):
        valor = int(input(f"Ingrese el valor para la posición ({fila}, {columna}): "))
        fila_mat.append(valor)
    mat.append(fila_mat)
    
print("Matriz ingresada:")
for fila in mat:
    print(fila)
```

```python
class Recursion:
    def facto(self, x):
        if x == 0:
            return 0
        
        return 2 * self.facto(x - 1) + x ** 2
    
# Programa principal
obj_recursion = Recursion()
print("f(1) = ", obj_recursion.facto(1))
print("f(2) = ", obj_recursion.facto(2))
print("f(3) = ", obj_recursion.facto(3))
print("f(4) = ", obj_recursion.facto(4))
print("f(5) = ", obj_recursion.facto(5))
```
