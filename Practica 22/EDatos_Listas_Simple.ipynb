Ademas implementaremos las operaciones basicas sobre listas (Insertar_Nodo_alfinal(), Insertar_Nodo_Ciclo, Insertar_Nodo_Ppio(), Imprimr_Informacion(), ListavaciaEtc(), etc...)

## Paso 1: Crear la clase Node y clase lista.

```python
class node:
    def __init__(self, data, next_node=None):
        self.data = data
        self.next_node = next_node

class lista:
    def __init__(self, name="lista"):
        self.name = name
        self.first_node = None
        self.last_node = None

    def is_empty(self):
        return self.first_node is None

    def size(self):
        count = 0
        current = self.first_node
        while current is not None:
            count += 1
            current = current.next_node
        return count

    def insertar_atras(self, insertitem):
        if self.is_empty():
            self.first_node = node(insertitem)
            self.last_node = self.first_node
        else:
            nuevo = node(insertitem)
            self.last_node.next_node = nuevo
            self.last_node = nuevo

    def insertar_frente(self, insertitem):
        if self.is_empty():
            self.first_node = node(insertitem)
            self.last_node = self.first_node
        else:
            nuevo = node(insertitem, self.first_node)
            self.first_node = nuevo

    def insertar_medio(self, insertitem):
        if self.is_empty():
            self.first_node = node(insertitem)
            self.last_node = self.first_node
            return

        tamano = self.size()
        
        if tamano % 2 != 0:
            pasos = (tamano // 2) + 1
        else:
            pasos = tamano // 2

        # Avanzamos hasta el nodo previo al punto de inserción
        current = self.first_node
        for _ in range(pasos - 1):
            current = current.next_node

        nuevo = node(insertitem, current.next_node)
        current.next_node = nuevo
        
        # Si se insertó al final del todo, actualizamos last_node
        if nuevo.next_node is None:
            self.last_node = nuevo

    def print(self):
        if self.is_empty():
            print("Empty", self.name)
            return
        print("El ", self.name, "es: ", end="")
        current = self.first_node
        while current is not None:
            print(current.data, end=" ")
            current = current.next_node
        print("\n")
```

## Paso 2 Crea un objeto de tipo lista.

```python
lista=lista()
```

## Paso 3: Inserta Objetos a la lista con las funciones atras, enfrente y en medio

```python
lista.insertar_atras(10)
lista.print()
lista.insertar_atras(234)
lista.print()
```

```python
lista.insertar_frente(24)
lista.print()
lista.insertar_frente(50)
lista.print()
```

```python
lista.insertar_medio(100)
lista.print()
lista.insertar_medio(200)
lista.print()
```
