```python
class Stack:
    def __init__(self, size):
        self.stck = [None] * size
        self.top = 0

    def push(self, ch):
        if self.top == len(self.stck):
            print("---Arreglo Lleno---")
            return
        self.stck[self.top] = ch
        self.top += 1

    def pop(self):
        if self.top == 0:
            print("---Arreglo Vacío---")
            return None
        self.top -= 1
        return self.stck[self.top]

    def full(self):
        return self.top == len(self.stck)

    def empty(self):
        return self.top == 0

    def capacity(self):
        return len(self.stck)

    def getNum(self):
        return self.top
    
# Codigo principal
pila = Stack(5)
for x in [10, 20, 30, 40, 50]:
    print("Push:", x)
    pila.push(x)

print("Capaci pila:", pila.capacity())
print("Num elem:", pila.getNum())
print("Pila llena:", pila.full())
pila.push(60)
while not pila.empty():
    print("Pop:", pila.pop())

print("Pila vacia:", pila.empty())
pila.pop()
```

```python
class Fila:
    def __init__(self, size):
        self.fila = [None] * size
        self.frente = 0
        self.final = 0
        self.num = 0

    def insert(self, x):
        if self.full():
            print("---Arreglo Lleno---")
            return
        self.fila[self.final] = x
        self.final = (self.final + 1) % len(self.fila)
        self.num += 1

    def delete(self):
        if self.empty():
            print("---Arreglo Vacío---")
            return None
        elemento = self.fila[self.frente]
        self.fila[self.frente] = None
        self.frente = (self.frente + 1) % len(self.fila)
        self.num -= 1
        return elemento

    def full(self):
        return self.num == len(self.fila)

    def empty(self):
        return self.num == 0

    def capacity(self):
        return len(self.fila)

    def getNum(self):
        return self.num
    
fila = Fila(5)
for x in [10, 20, 30, 40, 50]:
    print("Insertar:", x)
    fila.insert(x)

print("Capaci fila:", fila.capacity())
print("Num elem:", fila.getNum())
print("Fila llena:", fila.full())

fila.insert(60)
print("Borrar:", fila.delete())
print("Borrar:", fila.delete())
for x in [60, 70]:
    print("Insertar:", x)
    fila.insert(x)
while not fila.empty():
    print("Borrar:", fila.delete())

print("Fila vacia:", fila.empty())
fila.delete()
```

```python
class node:
    def __init__(self, data, next_node=None):
        self.data = data
        self.next_node = next_node

class lista:
    def __init__(self, name="lista"):
        self.name = name
        self.first_node = None
        self.last_node = None

    def is_empty(self):
        return self.first_node is None

    def size(self):
        count = 0
        current = self.first_node
        while current is not None:
            count += 1
            current = current.next_node
        return count

    def insertar_atras(self, insertitem):
        if self.is_empty():
            self.first_node = node(insertitem)
            self.last_node = self.first_node
        else:
            nuevo = node(insertitem)
            self.last_node.next_node = nuevo
            self.last_node = nuevo

    def insertar_frente(self, insertitem):
        if self.is_empty():
            self.first_node = node(insertitem)
            self.last_node = self.first_node
        else:
            nuevo = node(insertitem, self.first_node)
            self.first_node = nuevo

    def insertar_medio(self, insertitem):
        if self.is_empty():
            self.first_node = node(insertitem)
            self.last_node = self.first_node
            return

        tamano = self.size()

        if tamano % 2 != 0:
            pasos = (tamano // 2) + 1
        else:
            pasos = tamano // 2

        current = self.first_node
        for _ in range(pasos - 1):
            current = current.next_node

        nuevo = node(insertitem, current.next_node)
        current.next_node = nuevo

        if nuevo.next_node is None:
            self.last_node = nuevo

    def print(self):
        if self.is_empty():
            print("Vacio", self.name)
            return
        print("La ", self.name, "es: ", end="")
        current = self.first_node
        while current is not None:
            print(current.data, end=" ")
            current = current.next_node
        print("\n")
        
lista = lista()
lista.print()
lista.insertar_atras(10)
lista.print()
lista.insertar_atras(234)
lista.print()
lista.insertar_frente(24)
lista.print()
lista.insertar_frente(50)
lista.print()
lista.insertar_medio(99)
lista.print()
lista.insertar_atras(75)
lista.print()
print("Número de nodos en la lista:", lista.size())
```
