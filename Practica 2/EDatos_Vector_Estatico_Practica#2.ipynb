### Las estructuras de datos lineales 
### organizan los elementos de manera secuencial, donde cada elemento tiene un único predecesor y un único sucesor,
### facilitando el acceso y la manipulación de los datos.

## ¿Qué es un vector?
### Un vector es una estructura que permite guardar varios datos del mismo tipo.

## ¿Qué es un vector estático?
### Un vector estático es un vector cuyo tamaño se define desde el principio y no puede cambiar durante la ejecución.

## Practica No.2
### En esta práctica implementaremos un programa en python utilizando arreglos estaticos (estructuras de datos lineales estaticas) para calcular el promedio de 5 calificaciones 

## Paso 1: Importar las librerias

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Paso 2: Arreglo estatico de 5 posiciones

```python
calificaciones=[0]*5
calificaciones
```

## Paso 3: Captura las 5 calificaciones

```python
for i in range(5):
    calificaciones[i]=int(float(input("Ingrese la calificación del estudiante {}: ".format(i+1))))
```

## Paso 4: Calcular la suma

```python
suma = 0
for i in range (5):
    suma += calificaciones[i]
```

## Paso 5: Calcular el promedio

```python
promedio = suma / 5
```

## Paso 6: Mostrar resultados

```python
print("\nCalificaciones:")
print(calificaciones)
print("Suma:", suma)
print("Promedio:", promedio)
```
