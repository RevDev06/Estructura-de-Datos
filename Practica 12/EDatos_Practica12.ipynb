## 3.1 Pilas
### Definición: Una pila es una colección ordenada de elementos de la cual nuevos elementos, de la cual nuevos elementos seran insertado sy borrados por un extremo, llamado el TOP de la pila. Podemos diselar una pila con un arreglo estatico mas una filosofia LIFO (Last Input, First Output).

## Practica #12 Implementación de una Pila utilizando POO.
### En esta practica implementaremos las operaciones basicas de una pila (metodo push(x) y pop()), utilizando POO.

## Paso 1: Importar las librerias necesarias

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Paso 2: Implementamos la Pila utilizando POO.

```python
class pila:
    def __init__(self):
        self.top = -1
        self.s=[0]*3
        print("Pila inicializada")
        print("")

    def push(self, i):
        if self.top == 2:
            print("Pila Overflow")
            print("")
        else:
            self.top += 1
            self.s[self.top] = i
            print("Elemento agregado a la pila: ", i)
            print("")

    def pop(self):
        if self.top == -1:
            print("Pila Underflow")
            print("")
        else:
            print("Elemento eliminado de la pila: ", self.s[self.top])
            valor = self.s[self.top]
            self.top -= 1
            print("")
            return valor

    def recuperar(self, indice=None, valores=None):
        if indice is None:
            indice = self.top
            valores = []
        if indice == -1:
            print("Se recupero:", valores)
            return
        valores.append(self.s[indice])
        self.recuperar(indice - 1, valores)
```

## Paso 3. Crear objeto de la clase Pila.

```python
objeto_a = pila()
objeto_b = pila()

print("a push 20")
objeto_a.push(20)
print("b push 100")
objeto_b.push(100)
print("a push 30")
objeto_a.push(30)
print("a push 40")
objeto_a.push(40)
print("a push 50")
objeto_a.push(50)

print()
print("Recuperar en a:")
objeto_a.recuperar()
print()

print("b pop 100")
objeto_b.pop()
print("b pop 200")
objeto_b.pop()
```
