## 3.3.2 Listas Doblemente Enlazadas.

### Una lista doblemente enlazada (Doubly Linked List) es una estructura de datos lineal formada por una colección de nodos, donde cada nodo, contiene tres elementos:

### 1. Dato: Almacena la información.

### 2. Puntero anterior (anterior): Referencia al nodo anterior.

### 3. Puntero siguiente (siguiente): Referencia al nodo posterior.

### La caracteristica principal es que permite recorrer la lista en dos direcciones: hacia delante y hacia atras.

## Practica #24 Clase NodoDouble
### En esta practica implementaremos la clase nodo doble que define los tipos de objetos que seran referenciados en una lista doblemente enlazada.

## Paso 1: Crear la clase NodoDoble y ListaDoble.

```python
class NodeDouble:
    def __init__(self, data, left=None, right=None):
        self.data = data
        self.left = left
        self.right = right

class ListaDoble:
    def __init__(self, name="lista doble"):
        self.name = name
        self.first_node = None
        self.last_node = None

    def is_empty(self):
        return self.first_node is None

    def size(self):
        count = 0
        current = self.first_node
        while current is not None:
            count += 1
            current = current.right
        return count

    def insertar_frente(self, insertitem):
        if self.is_empty():
            self.first_node = NodeDouble(insertitem)
            self.last_node = self.first_node
        else:
            nuevo = NodeDouble(insertitem, left=None, right=self.first_node)
            self.first_node.left = nuevo
            self.first_node = nuevo

    def insertar_atras(self, insertitem):
        if self.is_empty():
            self.first_node = NodeDouble(insertitem)
            self.last_node = self.first_node
        else:
            nuevo = NodeDouble(insertitem, left=self.last_node, right=None)
            self.last_node.right = nuevo
            self.last_node = nuevo

    def insertar_medio(self, insertitem):
        if self.is_empty():
            self.first_node = NodeDouble(insertitem)
            self.last_node = self.first_node
            return

        tamano = self.size()
        if tamano % 2 != 0:
            pasos = (tamano // 2) + 1
        else:
            pasos = tamano // 2

        current = self.first_node
        for _ in range(pasos - 1):
            current = current.right

        siguiente = current.right
        nuevo = NodeDouble(insertitem, left=current, right=siguiente)
        current.right = nuevo

        if siguiente is not None:
            siguiente.left = nuevo
        else:
            self.last_node = nuevo

    def eliminar_frente(self):
        if self.is_empty():
            print("La lista está vacía.")
            return None

        removed_data = self.first_node.data
        if self.first_node == self.last_node:
            self.first_node = None
            self.last_node = None
        else:
            self.first_node = self.first_node.right
            self.first_node.left = None

        return removed_data

    def eliminar_atras(self):
        if self.is_empty():
            print("La lista está vacía.")
            return None

        removed_data = self.last_node.data
        if self.first_node == self.last_node:
            self.first_node = None
            self.last_node = None
        else:
            self.last_node = self.last_node.left
            self.last_node.right = None

        return removed_data

    def eliminar_medio(self):
        if self.is_empty():
            print("La lista está vacía.")
            return None

        if self.first_node == self.last_node:
            removed_data = self.first_node.data
            self.first_node = None
            self.last_node = None
            return removed_data

        tamano = self.size()
        if tamano % 2 != 0:
            posicion_eliminar = (tamano // 2) + 1
        else:
            posicion_eliminar = tamano // 2

        if posicion_eliminar == 1:
            return self.eliminar_frente()
        if posicion_eliminar == tamano:
            return self.eliminar_atras()

        current = self.first_node
        for _ in range(posicion_eliminar - 1):
            current = current.right

        removed_data = current.data
        nodo_anterior = current.left
        nodo_siguiente = current.right

        nodo_anterior.right = nodo_siguiente
        if nodo_siguiente is not None:
            nodo_siguiente.left = nodo_anterior

        return removed_data

    def print(self):
        if self.is_empty():
            print("La lista ", self.name, " está vacía")
            return
        print("La ", self.name, "es: ", end="")
        current = self.first_node
        while current is not None:
            print(current.data, end=" ")
            current = current.right
        print("\n")
```

## Paso 2: Crear un objeto de la clase ListaDoble llamada lista de tipo lista.

```python
lista = ListaDoble("Lista Doble")
```

## Paso 3: Inserta Objetos a la lista con las funciones atras, enfrente y en medio y enlazandolos entre si.

```python
lista.insertar_frente(10)
lista.print()
lista.insertar_medio(15)
lista.print()
lista.insertar_atras(30)
lista.print()
lista.insertar_frente(20)
lista.print()
lista.insertar_medio(25)
lista.print()
lista.insertar_atras(40)
lista.print()
```

## Paso 4: Eliminar los Objetos a la lista de atras, enfrente y en medio y arreglando los nodos enlazados para que sigan enlazados y no exista un vacio entre ellos.

```python
lista.print()   
lista.eliminar_frente()
lista.print()
lista.eliminar_medio()
lista.print()
lista.eliminar_atras()
lista.print()
```
