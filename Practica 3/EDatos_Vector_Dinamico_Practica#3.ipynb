## Las estructuras de datos lineales 
### organizan los elementos de manera secuencial, donde cada elemento tiene un único predecesor y un único sucesor,
### facilitando el acceso y la manipulación de los datos.

## ¿Qué es un vector?
### Un vector es una estructura que permite guardar varios datos del mismo tipo.

## ¿Qué es un vector dinámico?
### Un vector dinámico es un vector cuyo tamaño se define desde el principio y no puede cambiar durante la ejecución.

# Practica 3:
## En esta practica implementaremos un programa en python utilizando arreglos dinamicos (estructuras de datos lineales dianmicas) para calcular el promedio de n calificaciones

## Paso 1: Importar librerias

```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
```

## Paso 2: Capturar la cantidad de numeros

```python
n = int(input("Ingrese la cantidad de números que desea ingresar: "))
```

## Paso 3: Crea una lista:

```python
numeros=[]
```

## Paso 4: Captura los números

```python
for i in range(n):
    numero = int(input("Ingrese un número: "))
    numeros.append(numero)
```

## Paso 5: Calcular el promedio

```python
suma = 0 
for numero in numeros:
    suma += numero

promedio = suma / n
```

## Paso 6: Mostrar los resultados

```python
print("Los números ingresados son: ", numeros)
print("El promedio es: ", promedio)
```
