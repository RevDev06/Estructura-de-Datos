## 3.1 Pilas-
### Definición: Una pila es una colección ordenada de elementos de cual nuevos elementos seran insertados y borrados por un extremo llamado el TOP de la pila. Podemos diseñar una pila con un arreglo estatico mas una filosofia  LIFO (Last First Output)

## Practica 14 Implementación de una fila utilizando POO.
### En esta practica implentamos varios tipos de datos en los que podemos poner en una pila.

## Paso 1 importar las librerias necesarias.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Paso 2: Implementamos la pila utilizando POO

```python
class Stack:
    def __init__(self, size):
        self.stck = [None] * size
        self.top = 0
        
    def push(self, ch):
        if self.top == len(self.stck):
            print("---Arreglo Lleno---")
            return
        self.stck[self.top] = ch
        self.top += 1
    
    def pop(self):
        if self.top == 0:
            print("---Arreglo Vacío---")
            return None
        self.top -= 1
        return self.stck[self.top]
    
    def full(self):
        return self.top == len(self.stck)
    
    def empty(self):
        return self.top == 0
    
    def capacity(self):
        return len(self.stck)
    
    def getNum(self):
        return self.top
```

## Paso 3 implementamos una pila de números enteros

```python
pila_enteros = Stack(5)
pila_enteros.push(10)
pila_enteros.push(20)
pila_enteros.push(30)

print("Enteros, del tope hacia abajo:")
while not pila_enteros.empty():
    print(pila_enteros.pop())
```

## Paso 4: Implementamos una pilda de caracteres.

```python
# Pila de caracteres
pila_caracteres = Stack(5)
pila_caracteres.push('A')
pila_caracteres.push('B')
pila_caracteres.push('C')

print("\nCaracteres, del tope hacia abajo:")
while not pila_caracteres.empty():
    print(pila_caracteres.pop())
```

## Paso 5: Implementamos pila de objetos.

```python
class Producto:
    def __init__(self, nombre, precio):
        self.nombre = nombre
        self.precio = precio

    def __repr__(self):
        return f"Producto(nombre={self.nombre!r}, precio={self.precio})"

pila_objetos = Stack(3)
pila_objetos.push(Producto("Cuaderno", 45.5))
pila_objetos.push(Producto("Lapiz", 8.0))
pila_objetos.push(Producto("Mochila", 350.0))

print("\nObjetos, del tope hacia abajo:")
while not pila_objetos.empty():
    print(pila_objetos.pop())
```
