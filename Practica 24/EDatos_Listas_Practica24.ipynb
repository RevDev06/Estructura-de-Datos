## 3.3.2 Listas Doblemente Enlazadas.

### Una lista doblemente enlazada (Doubly Linked List) es una estructura de datos lineal formada por una colección de nodos, donde cada nodo, contiene tres elementos:

### 1. Dato: Almacena la información.

### 2. Puntero anterior (anterior): Referencia al nodo anterior.

### 3. Puntero siguiente (siguiente): Referencia al nodo posterior.

### La caracteristica principal es que permite recorrer la lista en dos direcciones: hacia delante y hacia atras.

## Practica #24 Clase NodoDouble
### En esta practica implementaremos la clase nodo doble que define los tipos de objetos que seran referenciados en una lista doblemente enlazada.

## Paso 1: Crear la clase Nodo.

```python
class NodeDouble:
    def __init__(self, data, left=None, right=None):
        self.data = data
        self.left = left
        self.right = right
```

## Paso 2: Crear un objeto de la clase NodoDouble llamada First.

```python
print("Crea a First")
First = NodeDouble(10)

print()
print("Recuperar a First")
print()

print("Direccion de First: ", First)
print()
print("Valor Info de First: ", First.data)
print()
print("Valor left de First: ", First.left)
print("Valor right de First: ", First.right)
```

## Paso 3: Crea un objeto de la calse nodo llamada Second

```python
print()
print("Crea a Second")
print()

Second = NodeDouble(20)

print()
print("Recuperar a Second")
print()

print("Direccion de Second: ", Second)
print()
print("Valor Info de Second: ", Second.data)
print()
print("Valor left de Second: ", Second.left)
print("Valor right de Second: ", Second.right)
```

## Paso 4: Ligar el prinmer nodo con el segundo nodo.

```python
print()
print("Conecta First con Second")
print()

First.right = Second
Second.left = First

print()
print("Direccion de Second: ")
print()

print("Dirección de Second: ", Second)
print("Direccion de First.right: ", First.right)
print()
print("Dirección de First: ", First)
print("Direccion de Second.left: ", Second.left)
```
