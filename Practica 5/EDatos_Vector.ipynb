### Las estructuras de datos lineales 
### organizan los elementos de manera secuencial, donde cada elemento tiene un único predecesor y un único sucesor,
### facilitando el acceso y la manipulación de los datos.

## ¿Qué es un vector?
### Un vector es una estructura que permite guardar varios datos del mismo tipo.

## ¿Qué es un vector dinámico?
### Un vector dinámico es un vector cuyo tamaño se define desde el principio y no puede cambiar durante la ejecución.

## ¿Qué es una matriz?
### Una matriz estatica es una estructura de datos que tiene un número fijo de filas y columnas, definido desde el momento en que se crea.
