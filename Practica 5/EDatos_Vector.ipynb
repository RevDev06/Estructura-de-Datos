### Las estructuras de datos lineales 
### organizan los elementos de manera secuencial, donde cada elemento tiene un único predecesor y un único sucesor,
### facilitando el acceso y la manipulación de los datos.

## ¿Qué es un vector?
### Un vector es una estructura que permite guardar varios datos del mismo tipo.

## ¿Qué es un vector dinámico?
### Un vector dinámico es un vector cuyo tamaño se define desde el principio y no puede cambiar durante la ejecución.

## ¿Qué es una matriz?
### Una matriz estatica es una estructura de datos que tiene un número fijo de filas y columnas, definido desde el momento en que se crea.

## Practica 5
### En esta practica diseñaremos un programa en python para implementar un arreglo estático de dos dimensiones (Matriz de 3 filas x 4 columnas )

## Paso 1: Immportar las librerias

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Paso 2: Arreglo estático de 2 dimensiones

```python
Filas = 3
Columnas = 4
```

## Paso 3: Crear la matriz con valores iniciales de 0

```python
matriz = [[0 for j in range(Columnas)] for i in range(Filas)]
matriz
```

## Paso 4: Capturamos los datos en la matriz

```python
for i in range(Filas):
    for j in range(Columnas):
        matriz[i][j] = int(input(f"Ingrese el valor para la posición ({i}, {j}): "))
        
```

## Paso 5: Mostrar el contenido de la estructura de datos (Matriz)

```python
print("\nMatriz:")
for i in range(Filas):
    for j in range(Columnas):
        print(matriz[i][j], end="\t")
    print()
```
