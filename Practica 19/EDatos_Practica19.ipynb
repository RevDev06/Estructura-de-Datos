## Practica # 19. Implementación de una Fila circular POO para almacenar objetos.

## En esta practica implementaremos por medio de una fila ciruclar, modicaremos diferentes objetos.

## Paso 1: Crear la clase persona 

```python
## Definición de la clase Persona
class Persona:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad
        
    def __str__(self):
        return f"Nombre: {self.nombre}, Edad: {self.edad}"
    
    # __repr__ permite que el objeto se imprima legiblemente dentro de listas: print(fila.queue)
    def __repr__(self):
        return f"Persona('{self.nombre}', {self.edad})"
```

Paso 2: Crear la fila circular

```python

class Queue:
    def __init__(self, size):
        if not isinstance(size, int) or size <= 0:
            raise ValueError("El tamaño debe ser un entero positivo")

        self.queue = [None] * size
        self.front = 0
        self.rear = 0
        self.num_elements = 0

    def enqueue(self, persona):
        # Validamos que el objeto sea instancia de la clase Persona
        if not isinstance(persona, Persona):
            raise TypeError("Solo se pueden insertar objetos de la clase Persona")

        if self.full():
            print("El arreglo está lleno")
            return

        self.queue[self.rear] = persona
        self.rear = (self.rear + 1) % len(self.queue)
        self.num_elements += 1

    def dequeue(self):
        if self.empty():
            print("El arreglo no tiene elementos")
            return None

        persona = self.queue[self.front]
        self.queue[self.front] = None
        self.front = (self.front + 1) % len(self.queue)
        self.num_elements -= 1
        return persona

    def full(self):
        return self.num_elements == len(self.queue)

    def empty(self):
        return self.num_elements == 0

    def capacity(self):
        return len(self.queue)

    def getnum(self):
        return self.num_elements


```

## Paso 3: Insertamos los valores

```python
fila = Queue(5)
fila.enqueue(Persona("Ana", 20))
fila.enqueue(Persona("Luis", 25))
fila.enqueue(Persona("María", 30))
fila.enqueue(Persona("Carlos", 35))
print("Paso 2:", fila.queue)
```

## Paso 4: Sacamos los valores de la fila ciruclar.

```python
fila.dequeue()
fila.dequeue()
fila.dequeue()
print("Paso 3:", fila.queue)
```

## Paso 5: Insertamos valores finales en la fila y imprimimos la fila.

```python
fila.enqueue(Persona("Elena", 40))
fila.enqueue(Persona("Pedro", 45))
fila.enqueue(Persona("Sofía", 50))
fila.enqueue(Persona("Miguel", 55))
print("Paso 4:", fila.queue)
```
