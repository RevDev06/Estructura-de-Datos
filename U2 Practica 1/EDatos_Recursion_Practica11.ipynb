# Unidad 2 Recursión
## 2.1.1 Definiciones recursivas
### Un objeto recursivo es aquel que aparece en la definicion de si mismo, en terminos simples, un metodo recursivo es un metodo que se llama asi mismo
### Se puede presentar de 2 formas distintas:
### La directa:
### Ocurre cuando un subprograma se llama a si mismo
### La indirecta: 
### Un subprograma llama a otro subprograma y en algun momento volverá a llamar al subprograma original
### Cuando se diseña una solucion recursiva para un problema se necesitan hacer dos cosas:
### - Definir el caso base.
### - Definir la regla par el caso general
## 2.1.2 Ejemplos de casos recursvos
## Ejemplo 1: Imprimir "Hola Mundo" 100 veces



```python
# Programa  recursivo oara imoprimir "Hello, World!" n veces
class RecusivoHelloWorld:
    # Metodo (funcion) recursivo
    def print_hello_world_n_times(n):
        
        # Caso base
        # Si n  es 0 o menor, no hacemos nada
        if n > 0:
            print("Hello, World!")
            # Llamada recursiva
            self.print_hello_world_n_times(n - 1)

    def __init__(self):
        pass

# Crear una instancia de la clase
objeto = RecusivoHelloWorld()
objeto.print_hello_world_n_times(5)  # Llamada inicial con n = 5
```

## Ejemplo 2: imprimir "Hola Mundo" 5 veces utilizando POO en Python

```python
# Programa recursivo utilizando clases y objetos

#Definicion de la clase
class RecursiveHelloWorld:
    
    def __init__(self, n):
        self.n = n
    
    # Metodo (funcion) recursivo
    def print_hello_world_n_times(self, n):
        
        # Caso base
        if self.n > 0:
            print("Hello, World!")
            self.n -= 1
            # Llamada recursiva
            self.print_hello_world_n_times(n - 1)

#Programa principal

#Crear un objeto de la clase
objeto = RecursiveHelloWorld(5)

#Llamar el metodo recursivo
objeto.print_hello_world_n_times(objeto.n)
```

## Ejemplo 3: calcula el factorial de numero N

```python
def factorial(n):
    
    if n==1:
        return 1
    else:
        return n * factorial(n-1)

#programa principal
print("Factorial de 5 es: ", factorial(5))
```

## Ejemplo 4: calcula factorial de un numero N con POO

```python
class factorial:
    def fact(self,n ):
        if n==1:
            return 1
        
        resultado = self.fact(n-1)*n
        return resultado
    
class recursion:
    def main():
        f= factorial()
        
        print("Factorial de 3 es: ", f.fact(3))
        print("Factorial de 4 es: ", f.fact(4))
        print("Factorial de 5 es: ", f.fact(5))

recursion.main()
```

## Ejemplo 5: Calcular funcion f(x) = 2f(x-1) + x**2, usando recursión

```python
def factorial(x):
    
    if x==0:
        return 0
    
    return 2 * factorial(x-1) + x**2

print("f(0) = ", factorial(0))
print("f(1) = ", factorial(1))
print("f(2) = ", factorial(2))
print("f(3) = ", factorial(3))
print("f(4) = ", factorial(4))
print("f(5) = ", factorial(5))
```

## Ejemplo 6: Calcular funcion f(x) = 2f(x-1) + x**2, usando recursión y POO

```python
class RecursionExample:
    def factorial(self, x):
        if x == 0:
            return 0
        
        return 2 * self.factorial(x - 1) + x ** 2
    
# Programa principal
recursion_example = RecursionExample()
print("f(0) = ", recursion_example.factorial(0))
print("f(1) = ", recursion_example.factorial(1))
print("f(2) = ", recursion_example.factorial(2))
print("f(3) = ", recursion_example.factorial(3))
print("f(4) = ", recursion_example.factorial(4))
print("f(5) = ", recursion_example.factorial(5))
```

## Ejemplo 7: Implemente una funcion recursiva, que capture un numero e imprima el mismo numero como salida

```python
def numero(n):
    if n==0:
        return "El numero es: 0" 
    else:
        return "El numero es: " + str(n)

print(numero(5))
print(numero(10))
print(numero(15))
```

## Ejemplo 8: implemente una funcion recursiva, que capture un numero de entrada e imprima el mismo numero utilizando POO

```python
class Impresion:
    def __init__(self, n): 
        self.n = n

    def numigu(self):
        print("El numero es: " + str(self.n))  
        

class Recursion:
    def main(self):
        f = Impresion(5)
        f.numigu()
        f = Impresion(10)
        f.numigu()
        f = Impresion(15)
        f.numigu()

Recursion().main()
```

## Ejemplo 9: Implementar un metodo recursivo para resolver la serie Fibonacci de un nunero "n".

```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)
i = int(input("Ingrese una posicipon para calcular el Fibonacci: "))
print("El numero de la posicion", i, "es", fibonacci(i))
```

## Ejercicio 10: Serie de Fibonacci con POO.

```python
class Fibonacci:
    def __init__(self, n):
        self.n = n

    def fib(self, n):
        if n <= 0:
            return 0
        elif n == 1:
            return 1
        else:
            return self.fib(n-1) + self.fib(n-2)
    
# Programa principal
class Recursion:
    def main(self):
        n = int(input("Ingrese una posición para calcular el Fibonacci: "))
        f = Fibonacci(n)
        print("El número de la posición", n, "es", f.fib(n))
        
Recursion().main()
```
