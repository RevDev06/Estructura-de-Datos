Definicion : Una lista simple en estructuras de datos, tambien llamada lista simplemente enlazada (en el idioma ingles, Singly Linked List), es una estructura que permite almacenar varios elementos conectados entre si mediante referencais o enlaces.

### Practica  # 20. Clase Nodo
En esta practica implementaremos la clase nodo que define los tipos de objetos que seran referenciados en una lista simple.

### Paso 1: Crea la clase Nodo.

```python
class Node:
    def __init__(self, data, nextNode=None):
        self.data = data
        self.nextNode = nextNode
```

### Paso 2: Crea un objeto clase nodo llamado First

```python
print("Crea a First")

First = Node(10)

print()
print("Recuperar a First")
print()

print("Dirección de First: ", First)
print()

print("Valor Info de First: ", First.data)
print()

print("Valor nextNode de First: ", First.nextNode)
```

### Paso 3: Crea un objeto de la clase nodo llamada Second

```python
print("Crea a Second")

Second = Node(20)

print()
print("Recuperar a Second")
print()

print("Dirección de Second: ", Second)
print()

print("Valor Info de Second: ", Second.data)
print()

print("Valor nextNode de Second: ", Second.nextNode)
```

### Paso 4: Ligar el primero nodo con el segundo nodo 

```python
print()
print("Conectar First con Second")
print()

First.nextNode = Second

print("Direcciones First y Second")
print()

print("Dirección de First: ", First)
print()

print("Direccion de First.nextNode: ", First.nextNode)
print()

print("Dirección de Second: ", Second)
print()

print("Direccion de Second.nextNode: ", Second.nextNode)
print()
```
