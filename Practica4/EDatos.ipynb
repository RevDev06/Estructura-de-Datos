# Estructuras de Datos Dinámicas
## Las estructuras de datos lineales 
### organizan los elementos de manera secuencial, donde cada elemento tiene un único predecesor y un único sucesor,
### facilitando el acceso y la manipulación de los datos.

## ¿Qué es un vector?
### Un vector es una estructura que permite guardar varios datos del mismo tipo.

## ¿Qué es un vector dinámico?
### Un vector dinámico es un vector cuyo tamaño se define desde el principio y no puede cambiar durante la ejecución.

# Practica 4:
## En esta practica implementaremos un programa en python utilizando arreglos dinamicos (estructuras de datos lineales dianmicas) para ordenar un vector de 5 números en forma ascendente

## Paso 1: Importar las librerias

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Paso 2: Crear un vector dinámico vacío 

```python
numeros = []
```

## Paso 3: Capturar la cantidad de números a ordenar.

```python
n = int(input("Ingrese la cantidad de números que desea ingresar: "))
```

## Paso 4: Capturar los números:

```python
for i in range(n):
    numero = float(input(f"Captura el número {i + 1}: "))
    numeros.append(numero)
```

## Paso 5: Mostrar el vector original

```python
print("Vector Original:")
print(numeros)
```
