## 3.2 Colas
### Definición: Una cola (Queue) es una estructura de datos que almacena elementos siguiendo pricnipios FIFO(First in, Firt Out), que signfica: el primero en entrar es el primero en salir.

### 3.2.2 Operaciones Basicas.

## Practica # 17. Implementación de una Fila utilizando POO para almacenar caracteres.

## En esta practica implementaremos las operaciones basicas de una fila (metodo Enquiuque(x), unquique(), empty(), full(), capacity(), etc.) utilizando POO.

## Paso 1: Creamos la clase persona 

```python
class  Persona:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad
        
    def __str__(self):
        return f"Nombre: {self.nombre}, Edad: {self.edad}"
```

## Paso 2: Crear la clase QUEUE

```python
class Queue:
    def __init__(self, size):
        self.queue=[None]*size
        self.front=0
        self.rear=0
    
    def enqueue(self, ch):
        if self.rear==len(self.queue):
            print("El arreglo está lleno")
            return

        self.queue[self.rear]=ch
        self.rear+=1
    
    def dequeue(self):
        if self.front==self.rear:
            print("El arreglo no tiene elementos")
            return None

        ch=self.queue[self.front]
        self.front+=1
        return ch

    #Verifica si el arreglo está lleno
    def full(self):
        return self.rear==len(self.queue)

    #Verifica si el arreglo está vacío
    def empty(self):
        return self.front==self.rear

    #Capacidad del arreglo
    def capacity(self):
        return len(self.queue)

    # Número de elementos en el arreglo
    def getnum(self):
        return self.rear-self.front
    
```

## Paso 3: Crear una fila de 10 objetos

```python
q1 = Queue(10)
```

## Paso 4: Insertar objetos en la fila

```python
print("Insertar objetos Persona en q1")
personas = [
    Persona("Juan", 25),
    Persona("María", 30),
    Persona("Pedro", 22),
    Persona("Ana", 28),
    Persona("Luis", 35)
]

for persona in personas:
    print("Objeto que se inserta en la cola:", persona)
    q1.enqueue(persona)
```

## Paso 5: Mostrar la informacion

```python
print("Capacidad de q1:", q1.capacity())
print("Número de elementos en q1:", q1.getnum())
```

## Paso 6: Sacar y mostrar informacion de la Fila

```python
print("\n Contenido de q1: ")
while not q1.empty():
    obj = q1.dequeue()
    print(obj)
```

## Paso /: Checar si la fila tiene información

```python
if q1.empty():
    print("\n q1 está vacía")
```
