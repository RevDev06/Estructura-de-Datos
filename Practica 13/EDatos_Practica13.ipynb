### Definición: Una pila es una colección ordenada de elementos de la cual nuevos elementos, de la cual nuevos elementos seran insertado sy borrados por un extremo, llamado el TOP de la pila. Podemos diselar una pila con un arreglo estatico mas una filosofia LIFO (Last Input, First Output).

## Practica #13 Implementación de una Pila utilizando POO para almacenar caracteres.
### En esta practica implementaremos las operaciones basicas de una pila (metodo push(), pop(), empty(), full(), capacity(), etc.), utilizando POO.

## Paso 1: Importar las librerias necesarias

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Paso 2: Implementamos la Pila utilizando POO.

```python
class Stack:
    def __init__(self, size):
        self.stck = [None] * size
        self.top = 0
        
    def push(self, ch):
        if self.top == len(self.stck):
            print("---Arreglo Lleno---")
            return
        self.stck[self.top] = ch
        self.top += 1
    
    def pop(self):
        if self.top == 0:
            print("---Arreglo Vacío---")
            return None
        self.top -= 1
        return self.stck[self.top]
    
    def full(self):
        return self.top == len(self.stck)
    
    def empty(self):
        return self.top == 0
    
    def capacity(self):
        return len(self.stck)
    
    def getNum(self):
        return self.top
```

## Paso 3. Programa Principal

```python
stck1 = Stack(10)
stck2 = Stack(10)
stck3 = Stack(10)
```

## Paso 4. Insertar "A" hasta "J" en el objeto stck1

```python
print("Push A hasta J en stck1.")
i = 0
while not stck1.full():
    ch = chr(ord('A') + i)
    stck1.push(ch)
    i += 1
    
if stck1.full():
    print("stck1 está llena.\n")
```

## Paso 5. Mostrar contenido del objeto 1 (stck1)

```python
print("Contenido de stck1:", end="")
while not stck1.empty():
    ch = stck1.pop()
    print(ch, end="")
    
print()

if stck1.empty():
    print("stck1 está vacía.\n")
```

## Paso 6. Insertar "A" hasta "J" en el objeto stck1 y pasar elementos del objeto 1 (stck1) al objeto 2 (stck2)

```python
print("Nuevamente insertar A hasta J en stck1.")
i = 0
while not stck1.full():
    ch = chr(ord('A') + i)
    stck1.push(ch)
    i += 1

print("Ahora sacar caracteres de stck1 y meterlos en stck2.")
while not stck1.empty():
    ch = stck1.pop()
    stck2.push(ch)
    
print("Contenido de stck2:", end="")
while not stck2.empty():
    ch = stck2.pop()
    print(ch, end="")

print("\n")

if stck1.full():
    print("stck1 está llena.\n")
```

## Paso 7. Insertar 5 elementos en el objeto 3 (stck3).

```python
print("Insertar 5 caracteres en stck3.")

for i in range(5):
    ch = chr(ord('A') + i)
    stck3.push(ch)
    
print("Capacidad de stck3:", stck3.capacity())
print("Numero de elementos en stck3:", stck3.getNum())
```
