## 3.2 Colas
### Definición: Una cola (Queue) es una estructura de datos que almacena elementos siguiendo pricnipios FIFO(First in, Firt Out), que signfica: el primero en entrar es el primero en salir.

### 3.2.2 Operaciones Basicas.

## Practica # 16. Implementación de una Fila utilizando POO para almacenar caracteres.

## En esta practica implementaremos las operaciones basicas de una fila (metodo Enquiuque(x), unquique(), empty(), full(), capacity(), etc.) utilizando POO.

## Paso 1: Creamos la clase Queue

```python
class Queue:
    def __init__(self, size):
        self.queue=[None]*size
        self.front=0
        self.rear=0
    
    def enqueue(self, ch):
        if self.rear==len(self.queue):
            print("El arreglo está lleno")
            return

        self.queue[self.rear]=ch
        self.rear+=1
    
    def dequeue(self):
        if self.front==self.rear:
            print("El arreglo no tiene elementos")
            return None

        ch=self.queue[self.front]
        self.front+=1
        return ch

    #Verifica si el arreglo está lleno
    def full(self):
        return self.rear==len(self.queue)

    #Verifica si el arreglo está vacío
    def empty(self):
        return self.front==self.rear

    #Capacidad del arreglo
    def capacity(self):
        return len(self.queue)

    # Número de elementos en el arreglo
    def getnum(self):
        return self.rear-self.front
    
```

## Paso 2: Programa principal

```python
q1=Queue(10)
q2=Queue(10)
```

## Paso 3: Insertar Caracteres de la "A" hasta la "J" en el objeto 1

```python
print("Enqueue A hasta J en q1")
i=0
while not q1.full():
    ch=chr(ord('A')+i)
    print("Caracter que se inserta en la cola: ", ch)
    q1.enqueue(ch)
    i+=1
if q1.full():
    print("q1 está llena")
    
```

## Paso 4: Mostrar el contenido del objeto 1.

```python
print("Contenido de q1:", end="")
while not q1.empty():
    ch=q1.dequeue()
    print(ch, end="")
print()

if q1.empty():
    print("q1 está vacía\n")
```

## Paso 5: Insertar 5 elementos en el objeto q2

```python
print("Insertar 5 caracteres en q2")
for i in range(5):
    ch=chr(ord('A')+i)
    q2.enqueue(ch)
```

## Paso 6: Informacion objeto q2

```python
print("Capacidad de q2: ", q2.capacity())
print("Número de elementos en q2: ", q2.getnum())
```
