## Las estructuras de datos lineales 
### organizan los elementos de manera secuencial, donde cada elemento tiene un único predecesor y un único sucesor,
### facilitando el acceso y la manipulación de los datos.

## ¿Qué es un vector?
### Un vector es una estructura que permite guardar varios datos del mismo tipo.

## ¿Qué es un vector estático?
### Un vector estático es un vector cuyo tamaño se define desde el principio y no puede cambiar durante la ejecución.

## Practica No.1 
### En esta práctica implemantaremos un programa en python utilizando arreglos estaticos

## Paso 1: Importar las librerias necesarias

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
```

## Paso 2: Definimos las constantes para los meses del año

```python
enero=0
febrero=1
marzo=2
abril=3
mayo=4
junio=5
julio=6
agosto=7
septiembre=8
octubre=9
noviembre=10
diciembre=11
```

## Paso 3: Arreglo estatico para almacenar los dias  de cada mes

```python
moth_days=[0]*12
moth_days
```

## Paso 4: Capturar el mes

```python
mes= int(input("captura el mes (0-11): "))
```

## Paso 5: Determinar los dias del mes

```python
if mes==enero:
    moth_days[mes]=31
elif mes==febrero:
    moth_days[mes]=28
elif mes==marzo:
    moth_days[mes]=31
elif mes==abril:
    moth_days[mes]=30
elif mes==mayo:
    moth_days[mes]=31
elif mes==junio:
    moth_days[mes]=30
elif mes==julio:
    moth_days[mes]=31
elif mes==agosto:
    moth_days[mes]=31
elif mes==septiembre:
    moth_days[mes]=30
elif mes==octubre:
    moth_days[mes]=31
elif mes==noviembre:
    moth_days[mes]=30
elif mes==diciembre:
    moth_days[mes]=31
```

## Paso 6: Mostrar los dias del mes capturado


```python
print("El mes capturado tiene", moth_days[mes], "dias.")
```
