## 3.2 Colas
### Definición: Una cola (Queue) es una estructura de datos que almacena elementos siguiendo pricnipios FIFO(First in, Firt Out), que signfica: el primero en entrar es el primero en salir.

### 3.2.2 Operaciones Basicas.

## Practica # 18. Implementación de una Fila circular POO para almacenar caracteres.

## En esta practica implementaremos por medio de una filaciruclar, modicaremos diferentes valores.

## Paso 1: Creamos la fila circular

```python
class Queue:
    def __init__(self, size):
        if not isinstance(size, int) or size <= 0:
            raise ValueError("El tamaño debe ser un entero positivo")

        self.queue = [None] * size
        self.front = 0
        self.rear = 0
        self.num_elements = 0

    def enqueue(self, number):
        if isinstance(number, bool) or not isinstance(number, (int, float)):
            raise TypeError("Solo se pueden insertar números")

        if self.full():
            print("El arreglo está lleno")
            return

        self.queue[self.rear] = number
        self.rear = (self.rear + 1) % len(self.queue)
        self.num_elements += 1

    def dequeue(self):
        if self.empty():
            print("El arreglo no tiene elementos")
            return None

        number = self.queue[self.front]
        self.queue[self.front] = None
        self.front = (self.front + 1) % len(self.queue)
        self.num_elements -= 1
        return number

    def full(self):
        return self.num_elements == len(self.queue)

    def empty(self):
        return self.num_elements == 0

    def capacity(self):
        return len(self.queue)

    def getnum(self):
        return self.num_elements
```

## Paso 2: Insertamos los valores

```python
fila=Queue(5) 
fila.enqueue(10)
fila.enqueue(20)
fila.enqueue(30)
fila.enqueue(40)
print(fila.queue)
```

## Paso 3: Sacamos los valores de la fila ciruclar.

```python
fila.dequeue()
fila.dequeue()
fila.dequeue()
print(fila.queue)
```

## Paso 4: Insertamos valores finales en la fila y imprimimos la fila.

```python
fila.enqueue(50)
fila.enqueue(60)
fila.enqueue(70)
fila.enqueue(80)
print(fila.queue)
```
