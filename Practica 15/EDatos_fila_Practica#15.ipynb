## Pilas
### Definición: Una pila es una colección ordenada de elementos de cual nuevos elementos seran insertados y borrados por un extremo llamado el TOP de la pila. Podemos diseñar una pila con un arreglo estatico mas una filosofia  LIFO (Last First Output)
## Filas
### Una fila es una colección ordenada de elementos en la cual los nuevos elementos son insertados por un extremo (llamado el final o rear) y los elementos son borrados o extraídos por el extremo opuesto (llamado el frente o front). Podemos diseñar una fila con un arreglo estático más una filosofía FIFO (First In, First Out / Primero en Entrar, Primero en Salir).

## Practica 15 Implementación de una fila utilizando POO.
### En esta practica implentaremos las operaciones basicas de una Fila(Metodo Qinsert(x) y metodo Qdelete() Utilizando POO)

## Paso 1 importar las librerias necesarias.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Paso 2: Creamos Fila con sus metodos y arreglos.

```python
class Fila:
    def __init__(self, size):
        self.fila = [None] * size
        self.frente = 0
        self.final = 0
        self.num = 0

    def Qinsert(self, x):
        if self.full():
            print("---Arreglo Lleno---")
            return
        self.fila[self.final] = x
        self.final = (self.final + 1) % len(self.fila)
        self.num += 1

    def Qdelete(self):
        if self.empty():
            print("---Arreglo Vacío---")
            return None
        elemento = self.fila[self.frente]
        self.fila[self.frente] = None
        self.frente = (self.frente + 1) % len(self.fila)
        self.num -= 1
        return elemento

    def full(self):
        return self.num == len(self.fila)

    def empty(self):
        return self.num == 0

    def capacity(self):
        return len(self.fila)

    def getNum(self):
        return self.num

```

## Paso 3: Creamos Fila de enteros

```python
fila_enteros = Fila(5)
fila_enteros.Qinsert(10)
fila_enteros.Qinsert(20)
fila_enteros.Qinsert(30)

print("Enteros, del primero al último:")
while not fila_enteros.empty():
    print(fila_enteros.Qdelete())
```

## Paso 4: Creamos fila de caracteres.

```python
fila_caracteres = Fila(5)
fila_caracteres.Qinsert('A')
fila_caracteres.Qinsert('B')
fila_caracteres.Qinsert('C')

print("\nCaracteres, del primero al último:")
while not fila_caracteres.empty():
    print(fila_caracteres.Qdelete())
```
